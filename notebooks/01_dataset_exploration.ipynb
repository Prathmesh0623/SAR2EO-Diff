## Discover pairs and check basic stats

```python
DATASET_ROOT = '/kaggle/input/sen12ms'  # <-- update to your actual Kaggle dataset path

ds = SEN12MSPairedDataset(root=DATASET_ROOT, split='train', patch_size=128, subset_size=50)
print('Number of pairs (subset):', len(ds))
```

```python
sample = ds[0]
print('SAR shape:', sample['sar'].shape, sample['sar'].dtype)
print('EO shape:', sample['eo'].shape, sample['eo'].dtype)
print('SAR value range:', sample['sar'].min().item(), sample['sar'].max().item())
print('EO value range:', sample['eo'].min().item(), sample['eo'].max().item())
```

## Visualize SAR VV / VH and EO RGB side by side

**What to look for:** SAR should look grainy/speckled and grayscale-textural (no obvious "natural color" cues); EO should look like a normal aerial photo. If SAR channels look identical to each other, double check the band indexing in `SEN12MSPairedDataset._load_sar`.

```python
from src.data.preprocessing import denormalize_sar, denormalize_eo

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sar_np = sample['sar'].numpy()
eo_np = denormalize_eo(sample['eo'].numpy())

axes[0].imshow(denormalize_sar(sar_np[0]), cmap='gray'); axes[0].set_title('SAR VV'); axes[0].axis('off')
axes[1].imshow(denormalize_sar(sar_np[1]), cmap='gray'); axes[1].set_title('SAR VH'); axes[1].axis('off')
axes[2].imshow(np.transpose(eo_np, (1,2,0))); axes[2].set_title('EO RGB'); axes[2].axis('off')
plt.tight_layout(); plt.savefig('../results/figures/01_sar_eo_example.png', dpi=150)
plt.show()
```
